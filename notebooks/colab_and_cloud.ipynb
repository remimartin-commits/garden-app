```python
import nltk
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

import spacy
nlp = spacy.load('en_core_web_sm')
doc = nlp('Codebot uses Python for chatbots and RAG.')
print([t.lemma_ for t in doc if not t.is_stop])
```
